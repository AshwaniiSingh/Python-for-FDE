```python
import copy

original = [[1, 2, 3], [4, 5, 6]]
deep_copied = copy.deepcopy(original)

original[0][2] = 99
# Modifying a nested element
deep_copied[0][0] = 99

print("Original:", original)    # Output: [[1, 2, 99], [4, 5, 6]]
print("Deep Copy:", deep_copied)   # Output: [[99, 2, 3], [4, 5, 6]]
```

```python
import copy

original = [[1, 2], [3, 4]]
copied = copy.copy(original)

# Modifying the top-level structure of the copy
copied.append([5, 6])
copied.pop(0)

# Accessing the first sub-list and appending to it
#copied[0].append(99)

print(original)  # Output: [[1, 2, 99], [3, 4]] -> AFFECTED!
print(copied)    # Output: [[1, 2, 99], [3, 4]]
```

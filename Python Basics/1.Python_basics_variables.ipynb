### Write a Python program to count Uppercase, Lowercase, special characters, and numeric values in a given string.

```python
def count_cases(s):
    upper = lower = digits = special = 0
    for char in s:
        if char.isupper(): upper += 1
        elif char.islower(): lower += 1
        elif char.isdigit(): digits += 1
        elif not char.isspace(): special += 1
    return upper, lower, digits, special

print(count_cases("saWERiy098&@^%"))
```

```python
# Sample string with a trailing newline and an internal newline
original_text = '''Hello World!
This is a new line.
This is the third line.'''

print("--- Original Text ---")
print(repr(original_text))
print("---------------------\n")


# Method 1: Using rstrip() 
# Best for removing trailing newlines at the end of a string.
text_rstrip = original_text.rstrip("\n")
print("1. Using rstrip() (removes trailing only):")
print(repr(text_rstrip))
print()


# Method 2: Using replace()
# Best for removing ALL newlines anywhere in the string.
text_replace = original_text.replace("\n", " ")
print("2. Using replace() (removes all newlines):")
print(repr(text_replace))
print()


# Method 3: Using splitlines() and join()
# Best for multi-line strings to cleanly combine lines without extra spaces.
text_split_join = " ".join(original_text.splitlines())
print("3. Using splitlines() and join() (removes all newlines):")
print(repr(text_split_join))
```

```python
23.	Write a Python program to swap commas and dots in a string
●	Sample string: "32.054,23" >> Expected Output: "32,054.23"

# Sample string
sample_string = "32.054,23"

# Swap commas and dots
swapped_string = sample_string.translate(str.maketrans({",": ".", ".": ","}))

# Print the result
print(swapped_string)  # Output: 32,054.23

import re

# Sample string
sample_string = "32.054,23"

# Use a function to determine the replacement character
def swap(match):
    char = match.group(0)
    return "." if char == "," else ","

# Find all commas and dots and swap them
swapped_string = re.sub(r"[.,]", swap, sample_string)

# Print the result
print(swapped_string)  # Output: 32,054.23


# Sample string
sample_string = "32.054,23"

# Initialize an empty string to build the result
swapped_string = ""

# Loop through each character in the string
for char in sample_string:
    if char == ".":
        swapped_string += ","
    elif char == ",":
        swapped_string += "."
    else:
        swapped_string += char

# Print the result
print(swapped_string)  # Output: 32,054.23
```

```python
24.	Write a Python program to find the first repeated character in a given string

def first_repeated(s):
    seen = set()
    for char in s:
        if char in seen:
            return char
        seen.add(char)
    return None
```

```python
25.	Write a Python program to find the second most repeated word in a given string

from collections import Counter
def second_most_repeated(s):
    words = s.split()
    counts = Counter(words)
    if len(counts) < 2: return None
    return counts.most_common(2)[1][0]
```

```python
def manual_swap_case(input_string):
    result = []
    
    # Iterate through each character in the string
    for char in input_string:
        if char.isupper():
            result.append(char.lower())
        elif char.islower():
            result.append(char.upper())
        else:
            # Keep numbers, spaces, and symbols as they are
            result.append(char)
            
    # Join the character list back into a single string
    return "".join(result)

# Test the manual function
text = "Hello World 123!"
print("Manual Swapped:", manual_swap_case(text))
```

```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt
import os
```

```python
x = np.linspace(0,5,11)
x
y = x **3 
y
```

```python
plt.plot(x,y)
plt.title("y=x^2")
plt.xlabel("x")
plt.ylabel("ylabel")
```

```python
plt.subplot(2,2,1)
plt.plot(x,x*10)
plt.subplot(2,2,2)
plt.plot(x,x)
plt.subplot(2,2,3)
plt.plot(y,x)
plt.subplot(2,2,4)
plt.plot(x,y)
```

### Object Oriented Way

```python
fig = plt.figure()
axis1 = fig.add_axes([.1,.1,1,1])
axis1.plot(x,y)
axis2 = fig.add_axes([.3,.4,.3,.5])
axis2.plot(x,y)
plt.savefig("sqr.png")
print(os.getcwd())
```
